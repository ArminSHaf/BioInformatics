### EXercise 1

The melting temperature(Tm) is the temp at which one half of a particular DNA duplex will dissociate and become a simple strand of DNA. Primer length and sequence are of critical importance i designing the parameters of a successful amplific. The Tm of a nucleic asset duplex increases both with its length and with an increased C.G. content. A simple formula for calculating.
with an increased CG code
Tm = 4(G+C)+2(A+T);
Make an app in any environment or programming language which si able to sclalulcate the Tm based on an input sequence ; 
## S =ATCGCGTA 

```python
def calculate_tm(sequence):
    sequence = sequence.upper()
    a_count = sequence.count("A")
    t_count = sequence.count("T")
    g_count = sequence.count("G")
    c_count = sequence.count("C")
    return 4 * (g_count + c_count) + 2 * (a_count + t_count)

S = "ATCGCGTA"
tm = calculate_tm(S)

print(f"Sequence: {S}")
print(f"Tm: {tm}°C")
```

### Exercise 2

The actual Tm is influenced by the concentrations of Mg, K, and other components. An alternative formula is:

\[
    Tm = 81.5 + 16.6*log10([Na+]) + 0.41(%GC) - 600/length
\]

Implement an app that calculates the Tm of a DNA sequence using one of these formulas, or both.

Input = ACGCGTGCCA

```python
import math

def calculate_tm_advanced(sequence, na_concentration=1.0):
    sequence = sequence.upper()
    
    # Calculate GC percentage
    gc_count = sequence.count("G") + sequence.count("C")
    gc_percentage = (gc_count / len(sequence)) * 100
    
    # Calculate Tm
    tm_advanced = 81.5 + 16.6 * math.log10(na_concentration) + 0.41 * gc_percentage - (600 / len(sequence))
    
    return tm_advanced, gc_percentage

# Test with the new sequence
sequence_input = "ACGCGTGCCA"
tm_advanced, gc_percent = calculate_tm_advanced(sequence_input)

print(f"Sequence: {sequence_input}")
print(f"Length: {len(sequence_input)}")
print(f"GC Content: {gc_percent:.2f}%")
print(f"Tm (Advanced Formula): {tm_advanced:.2f}°C")
```

## Sliding Window Score Interpretation

The sliding window score is the **sum of PWM values** for each nucleotide at each position in the window.

| Score | Meaning |
|---|---|
| **Positive** | The window matches the motif **better than random** — this region is likely a real splice site |
| **Zero** | The window matches the motif **exactly as expected by chance** — no signal |
| **Negative** | The window matches the motif **worse than random** — this region is unlikely to be a splice site |

The **higher** the score, the stronger the match to the known motif pattern.
